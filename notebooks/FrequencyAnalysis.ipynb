```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats
from PIL import Image
```

```python
#读取原图进行频谱分析
original_image = Image.open('../images/sample_image.jpg').convert('L')
original_image_array = np.array(original_image)
#计算原图频谱
original_image_fft = np.fft.fft2(original_image_array)
original_image_fft_shifted = np.fft.fftshift(original_image_fft)
#显示原图和频谱
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(original_image_array, cmap='gray')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.title('Frequency Spectrum of Original Image')
plt.imshow(np.log(np.abs(original_image_fft_shifted) + 1), cmap='gray')
plt.axis('off')
plt.show()
```

```python
#读取高斯模糊图像进行频谱分析
gaussian_blurred_image = Image.open('../outputs/gaussian_blurred_image.jpg').convert('L')
gaussian_blurred_image_array = np.array(gaussian_blurred_image)
gaussian_blurred_image_fft = np.fft.fft2(gaussian_blurred_image_array)
gaussian_blurred_image_fft_shifted = np.fft.fftshift(gaussian_blurred_image_fft)
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.title('Gaussian Blurred Image')
plt.imshow(gaussian_blurred_image_array, cmap='gray')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.title('Frequency Spectrum of Gaussian Blurred Image')
plt.imshow(np.log(np.abs(gaussian_blurred_image_fft_shifted) + 1), cmap='gray')
plt.axis('off')
plt.show()
```
