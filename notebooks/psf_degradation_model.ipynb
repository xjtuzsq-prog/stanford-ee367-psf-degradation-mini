```python
#导入必要库
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats
from PIL import Image
```

```python
#读取图片并转化为灰度图
image=Image.open('../images/sample_image.jpg')
gray_image=np.array(image.convert('L'))
#进行归一化
normalized_image=gray_image/255.0
```

```python
#生成高斯PSF
def gaussian_psf(size, sigma):
    """生成高斯PSF"""
    x = np.arange(-size//2 + 1, size//2 + 1)
    y = np.arange(-size//2 + 1, size//2 + 1)
    x, y = np.meshgrid(x, y)
    psf = np.exp(-(x**2 + y**2) / (2 * sigma**2))
    psf /= psf.sum()  # 归一化
    return psf
```

```python
#生成defocus PSF
def defocus_psf(size, radius):
    """生成defocus PSF"""
    x = np.arange(-size//2 + 1, size//2 + 1)
    y = np.arange(-size//2 + 1, size//2 + 1)
    x, y = np.meshgrid(x, y)
    r = np.sqrt(x**2 + y**2)
    psf = (r <= radius).astype(float)
    psf /= psf.sum()  # 归一化
    return psf
```

```python
#用PSF卷积图像
def convolve_image(image, psf):
    """用PSF卷积图像"""
    from scipy.signal import convolve2d
    return convolve2d(image, psf, mode='same', boundary='wrap') 
```

```python
#用Gaussian PSF卷积图像
gaussian_psf_kernel = gaussian_psf(size=21, sigma=3)
blurred_image_gaussian = convolve_image(normalized_image, gaussian_psf_kernel)
#显示图像
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(normalized_image, cmap='gray')
plt.subplot(1, 2, 2)
plt.title('Gaussian Blurred Image')
plt.imshow(blurred_image_gaussian, cmap='gray')
plt.axis('off')
plt.show()

```

```python
#用defocus PSF卷积图像
defocus_psf_kernel = defocus_psf(size=21, radius=5)
blurred_image_defocus = convolve_image(normalized_image, defocus_psf_kernel)
#显示结果
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(normalized_image, cmap='gray')
plt.subplot(1, 2, 2)
plt.title('Defocus Blurred Image')
plt.imshow(blurred_image_defocus, cmap='gray')
plt.axis('off')
plt.show()
```
